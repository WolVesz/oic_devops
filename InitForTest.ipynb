```python
###  Rotate Password
from typing import Dict
from oic_devops.client import OICClient
from oic_devops.workflows.password_rotation import PasswordRotationWorkflow
from typing import Any, Dict, Iterable, List, Tuple, Optional
import pandas as pd
import os
from datetime import datetime
from oic_devops.workflows.base import BaseWorkflow, WorkflowResult

config_file = "C:/Users/jorher/OneDrive - Tri-State Generation and Transmission Association, Inc/Documents/GitHub/oic_devops/config.yaml"
client = OICClient(profile="dev-1", config_file=config_file)
wf = PasswordRotationWorkflow(client)
self = wf
wf.execute()
init_result = WorkflowResult()
init_result.success = True
```

```python
test_integrations_ids = [
    'PSSWRD_EVENT_PRODUCER|01.00.0000', #no connection dependency
    'PSSWRDRE_BASIC_AUTH_VBCSBO|01.00.0000', #Scheduler
    'PSSWRD_ADAPTERHCM_INT|01.00.0000', #Scheduler
    'PSSWRD_SOAPHCMBIP_INT|01.00.0000',#Scheduler    
    'PSSWRD_REST_OIC_INC|01.00.0000',#Scheduler
    'PSSWRD_ALWAYS_AT_DRAFT|01.00.0000', #Never active
    "PSSWRD_ALWAYS_NOSCHEDULE|01.00.0000", #No scheduler defined
    "PSSWRD_SUBSCRIBER_A|01.00.0000", #Application Event Subscriber
    "PSSWRD_SUBSCRIBER_B|01.00.0000", #Application Event Subscriber
    "PSSWRD_BASIC_AUTH_VBCSBO_EVENT|01.00.0000" #Event Subscriber
]
test_integration_ids_do_not_activate = [
    'PSSWRD_ALWAYS_AT_DRAFT|01.00.0000', #Never active
]
test_sch_integrations_ids = [
    'PSSWRDRE_BASIC_AUTH_VBCSBO|01.00.0000', #Scheduler
    'PSSWRD_ADAPTERHCM_INT|01.00.0000', #Scheduler
    'PSSWRD_SOAPHCMBIP_INT|01.00.0000',#Scheduler    
    'PSSWRD_REST_OIC_INC|01.00.0000',#Scheduler
    'PSSWRD_ALWAYS_AT_DRAFT|01.00.0000', #Never active
    "PSSWRD_ALWAYS_NOSCHEDULE|01.00.0000" #No scheduler defined
]

## Fake the original Schedule state to ACTIVE
integrations_original_status: Dict[str, Dict[str, Any]] = {}

for integration_id in test_integrations_ids:
    integrations_original_status[integration_id] ={"SCHEDULE":{"state":"ACTIVE"}}
for integration_id in test_integration_ids_do_not_activate:
    integrations_original_status[integration_id] ={"SCHEDULE":{"state":"CONFIGURED"}}

init_result.success = True


print('variables initialized.')
```

```python
### ACTIVATE
integration_ids: list[str] = test_integrations_ids.copy()

for integration_id in test_integration_ids_do_not_activate:
    if integration_id in integration_ids:
        integration_ids.remove(integration_id)
    
activate_wf = wf.activate_integrations(integration_ids)

print(f"resources: {len(activate_wf.resources['integration'])} success:{activate_wf.success}  message: {activate_wf.message}")
for resource in activate_wf.resources['integration']:
    print(resource, 
          "\taction:",activate_wf.resources['integration'][resource]['action'],
          "\tdatetime:", activate_wf.resources['integration'][resource]['datetime'],
          "\tsucess:",activate_wf.resources['integration'][resource]['success'],
         "\tstatus:",activate_wf.resources['integration'][resource]['status'],)

init_result.merge(activate_wf)
if not activate_wf.success:
    exit(1)    
```

```python
## Start Scheduler
    
activate_wf = wf.start_schedulers(schedule_integrations=test_sch_integrations_ids, integrations_original_status=integrations_original_status)

print(f"resources: {len(activate_wf.resources['integration'])} success:{activate_wf.success}  message: {activate_wf.message}")
for resource in activate_wf.resources['integration']:
    print(resource,
          "\taction:",activate_wf.resources['integration'][resource]['action'],
          "\tdatetime:", activate_wf.resources['integration'][resource]['datetime'],
          "\tsuccess:",activate_wf.resources['integration'][resource]['success'],
          "\tschedule_state:",activate_wf.resources['integration'][resource]['schedule_state'],
         "\nmessage:",activate_wf.resources['integration'][resource]['message'],"\n"
         )

init_result.merge(activate_wf)
if not activate_wf.success:
    exit(1)        
```
