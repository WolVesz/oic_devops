```python
test_integrations_ids = [
    'PSSWRDRE_BASIC_AUTH_VBCSBO|01.00.0000',
    'PSSWRD_ADAPTERHCM_INT|01.00.0000', 
    'PSSWRD_SOAPHCMBIP_INT|01.00.0000',    
    'PSSWRD_REST_OIC_INC|01.00.0000',
    "PSSWRD_ALWAYS_NOSCHEDULE|01.00.0000", #No scheduler defined
    'PSSWRD_ALWAYS_AT_DRAFT|01.00.0000', #No active    
]
## Fake the original Schedule state to ACTIVE
integrations_original_status: Dict[str, Dict[str, Any]] = {}

for integration_id in test_integrations_ids:
    integrations_original_status[integration_id] ={"SCHEDULE":{"state":"ACTIVE"}}

print("ready to roll")
```

```python
## Activate integrations
activate_wf = wf.activate_integrations(integrations_original_status)
    
```

```python
## Start Scheduler
self = wf
activate_wf = wf.start_schedulers(schedule_integrations=test_integrations_ids, integrations_original_status=integrations_original_status)

for resource in activate_wf.resources[self.RESOURCE_INTEGRATION]:
    print(resource,
          "\taction:",activate_wf.resources[self.RESOURCE_INTEGRATION][resource]['action'],
          "\tdatetime:", activate_wf.resources[self.RESOURCE_INTEGRATION][resource]['datetime'],
          "\tsuccess:",activate_wf.resources[self.RESOURCE_INTEGRATION][resource]['success'],
          "\tschedule_state:",activate_wf.resources[self.RESOURCE_INTEGRATION][resource]['schedule_state'],
         "\nmessage:",activate_wf.resources[self.RESOURCE_INTEGRATION][resource]['message'],"\n"
         )
```

```python
integrations_ids = [
    'PSSWRDRE_BASIC_AUTH_VBCSBO|01.00.0000',
    'PSSWRD_ADAPTERHCM_INT|01.00.0000', 
    'PSSWRD_SOAPHCMBIP_INT|01.00.0000',    
    'PSSWRD_REST_OIC_INC|01.00.0000',
    # 'PSSWRD_ALWAYS_AT_DRAFT|01.00.0000', #No active
    # "PSSWRD_ALWAYS_NOSCHEDULE|01.00.0000" #No scheduler defined
]

wait_wf  =wf.wait_for_schedules_state(
                schedule_integrations= integrations_ids,
                integrations_original_status=integrations_original_status,
                desired_state = "ACTIVE",
                timeout_sec = 4,
                poll_interval_sec = 5)

print(f"\nsuccess: {wait_wf.success}   message: {wait_wf.message}\n")
for error in wait_wf.errors:
    print(error)
```
