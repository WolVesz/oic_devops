```python
test_integrations_ids = [
    'PSSWRDRE_BASIC_AUTH_VBCSBO|01.00.0000',
    'PSSWRD_ADAPTERHCM_INT|01.00.0000', 
    'PSSWRD_SOAPHCMBIP_INT|01.00.0000',    
    'PSSWRD_REST_OIC_INC|01.00.0000',
    'PSSWRD_ALWAYS_AT_DRAFT|01.00.0000', #No active
    "PSSWRD_ALWAYS_NOSCHEDULE|01.00.0000" #No scheduler defined
]

## Fake the original Schedule state to ACTIVE
integrations_original_status: Dict[str, Dict[str, Any]] = {}

for integration_id in test_integrations_ids:
    integrations_original_status[integration_id] ={"SCHEDULE":{"state":"ACTIVE"}}
```

```python
scheduled_ids =test_integrations_ids
# 4) Stop schedules and wait until CANCELLED
stop_sched_wf = self.stop_schedulers(scheduled_ids, integrations_original_status)

print(f"success: {stop_sched_wf.success}   message: {stop_sched_wf.message}")
for resource in stop_sched_wf.resources[self.RESOURCE_INTEGRATION]:
    print(resource, "\taction:",stop_sched_wf.resources[self.RESOURCE_INTEGRATION][resource]['action'],
          "\tdatetime:",        stop_sched_wf.resources[self.RESOURCE_INTEGRATION][resource]['datetime'],
          "\t sucess:",stop_sched_wf.resources[self.RESOURCE_INTEGRATION][resource]['success'])
```
